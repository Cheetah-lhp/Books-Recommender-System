# Model comparison
Compare ranking and beyond-accuracy metrics from one reproducible run.

```python
import json, pandas as pd
from pathlib import Path
run_id = Path('../../artifacts/LATEST').read_text().strip()
report = json.loads(Path(f'../../artifacts/{run_id}/metrics.json').read_text())
comparison = pd.DataFrame(report['models']).T
comparison.sort_values('ndcg@20', ascending=False) if 'ndcg@20' in comparison else comparison
```
