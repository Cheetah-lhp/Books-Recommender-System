# Interaction EDA
Uses the same prepared Parquet contract as the training CLI.

```python
from books_recommender.config import load_config
from books_recommender.spark import create_spark
config = load_config('../../configs/demo.yaml')
spark = create_spark(config)
```

```python
from pathlib import Path
interactions = spark.read.parquet(str(Path(config['paths']['processed']) / 'interactions'))
interactions.groupBy('split').count().show()
interactions.select('strength', 'rating').summary().show()
```
