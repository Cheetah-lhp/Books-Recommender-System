```python
import os
import json
import csv
from pathlib import Path
```

```python
DIR = Path.cwd().parent / 'data'
```

```python
for file in os.listdir(DIR):
    print(f"\n--- Mẫu từ file {file} ---")

    if file.endswith('.json'):
        with open(DIR / file, 'r', encoding='utf-8') as f:
            print(json.dumps(json.loads(next(f)), indent=4, ensure_ascii=False))

    if file.endswith('.csv'):
        with open(DIR / file, 'r', encoding='utf-8') as f:
            reader = csv.DictReader(f)
            print(json.dumps(next(reader), indent=4, ensure_ascii=False))
```
