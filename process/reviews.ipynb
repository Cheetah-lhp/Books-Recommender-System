```python
from pathlib import Path
from pyspark.sql import SparkSession
from pyspark.sql.functions import count
```

```python
DATA_DIR = Path.cwd() / 'data'
```

```python
spark = SparkSession.builder \
    .appName("Goodreads") \
    .config("spark.driver.memory", "8g") \
    .getOrCreate()
```

```python
for json_file in DATA_DIR.glob('*.json'):
    df = spark.read.json(str(json_file))

    total_rows = df.count()
    coverage_exprs = [
        ((count(c) / total_rows) * 100).alias(c) for c in df.columns
    ]

    df_coverage = df.agg(*coverage_exprs)

    coverage_dict = df_coverage.collect()[0].asDict()

    print(f"\nĐang kiểm tra file: {json_file.name} ({total_rows:,} dòng)")
    print("-" * 55)
    
    for col_name, coverage in coverage_dict.items():
        cov_val = coverage if coverage is not None else 0.0
        print(f"  - {col_name:<35}: {cov_val:>6.2f}%")

    print("=" * 55)

    df.unpersist()
```
